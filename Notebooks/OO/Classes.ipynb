# Classes

# Classes 

- Correspondem a tipos de dados.
    - `Int`, `String`, `List`, ..., são todos classes.
- Mas classes podem representar dados mais complexos.
    - Compostos por várias informações inter-relacionadas.
    - *e.g.*, uma classe que representa um jogo de Xadrez.
        - Estado do tabuleiro.
        - De quem é a vez da próxima jogada.
        - ...
- Classes também permitem armazenar de forma coerente informações e funções que as manipulam.
- Dois elementos básicos: **propriedades** e **funções**.

# Declarando e utilizando uma classe

- Exemplo mínimo:

````kotlin
class Thing   // Declaração de uma classe chamada "Thing"

val oneThing = Thing()   // Instanciação da classe
println(oneThing)

val otherThing = Thing()
println(otherThing)

println(oneThing == otherThing)
````

- A classe `Thing` é vazia, portanto a impressão simplesmente mostra o **identificador** do objeto.
- Analogamente, a comparação só pode testar se os identificadores são iguais.
    - *i.e.*, se são **o mesmo objeto**, e não se são **objetos iguais**. 

# Propriedades

- Simplificadamente, **propriedades** correspondem às informações armazenadas na classe.
- Sintaticamente similares a valores / variáveis.
- Podem ser declaradas no corpo da classe:

````kotlin
class Circle {
    var radius = 0.0
    var area = 0.0
}
````

- Podem ser acedidas através do operador `.`:

````kotlin
val c1 = Circle()
c1.radius = 5.0
c1.area = c1.radius * c1.radius * c1.pi
println(c1.area)
````

# Construtor Primário

- Comumente, queremos já instanciar classes com valores iniciais para algumas das suas propriedades.
- Isto pode ser feito com **construtores**.
- Em Kotlin, o **construtor primário** pode ser definido entre parêntesis após o nome da classe:

````kotlin
class Circle (val radius: Double = 0.0) {
    val pi = 3.14159
    val area = radius*radius*pi
}
````

- **Nota**: propriedades declaradas no corpo da classe podem usar os valores passados no construtor na sua inicialização.
- Exemplo de instanciação:

````kotlin
val c1 = Circle(5.0)
//c1.radius = 5.0    // Feito pelo construtor
//c1.area = c1.radius * c1.radius * c1.pi  // Automaticamente calculado na instanciação
println(c1.area)
````

# Construtor Primário (II)

- Algumas vezes, podemos querer passar para o construtor valores importantes para a instanciação da classe, mas que não precisam ser armazenados a longo prazo.
- Exemplo: digamos que não seja importante guardar o raio do círculo.
    - Mas ainda é necessário na construção para o cálculo da área.
    - Simplesmente, omitimos o `val` da declaração do parâmetro do construtor.

````kotlin
class Circle (radius: Double = 0.0) {
    val pi = 3.14159
    val area = radius*radius*pi
}
````

# Composição de Classes

- Propriedades de uma classe não são limitadas a tipos básicos da linguagem.
- Uma propriedade pode ser de qualquer tipo/classe.

````kotlin
class Point(val x: Double, val y: Double)

val origin = Point(0.0, 0.0)

class MovingCircle(val radius: Double, var center: Point = origin){

    val pi = 3.14159
    val area = pi * radius * radius   // Computed in object creation and stored
}
````

- Aqui, `center` é do tipo `Point`.
    - Note, adicionalmente, que foi declarado como `var`, o que significa que pode ser alterado.

````kotlin
val mc = MovingCircle(5.0, origin)
mc.center = Point(1.0, 2.0) // Movemos o círculo para (1, 2)
````

# O Modificador `const`

- Nos exemplos anteriores, `pi` é declarado como uma propriedade de `Circle` / `MovingCircle`.
    - Cada instância irá armazenar um `Double` com este valor.
- Na prática, `pi` não é uma propriedade de um círculo em particular.
    - É uma propriedade geométrica de qualquer círculo.
- Alternativa melhor:

```Kotlin
const val PI = 3.14159
class MovingCircle(val radius: Double, var center: Point = origin){
    val area = PI * radius * radius
}
```

- `const` diz ao compilador para substituir `PI` pelo valor literal (`3.14159`) em **tempo de compilação**.

# Funções Membro

- Também chamadas **métodos**.
- Declaradas no corpo da classe.
- Exemplo de uma função `move()`:

````kotlin
val PI = 3.14159
class MovingCircle(val radius: Double, var center: Point = origin){
    val area = PI * radius * radius
    fun move(dx: Double, dy: Double) {
       center = Point(center.x + dx, center.y + dy)
   }
}
````

- Exemplo de uso:

````kotlin
val mc = MovingCircle(5.0, origin)
mc.move(1.2, 2.7)
println("Centro do círculo é em (${mc.center.x}, ${mc.center.y})")
````

# Restrições de Acesso

- Por vezes, propriedades e métodos de uma classe apenas são úteis internamente.
    - Não faz sentido que troços de código que utilizam a classe tenham acesso a estes.
- Por omissão, propriedades e métodos em Kotlin são **públicos**.
- Mas podemos torná-los privados com o modificador `private`.
- Exemplo:
    - Queremos representar um círculo que pode mudar de raio.
    - Mas só queremos expor a informação da área resultante.

# Restrições de Acesso: Exemplo

````kotlin
val PI = 3.14159
class GrowingCircle(private var radius: Double){
    var area = PI * radius * radius
    private fun updateArea() {
        area = PI * radius * radius  // Atualiza a área
    }
    fun grow() { 
        radius++
        updateArea()
    }
}
val gc = GrowingCircle(5.0)
//println(gc.radius)    // Erro!
println(gc.area)
gc.grow()
//gc.updateArea()  // Erro!
println("Nova área: ${gc.area}")
````

# Restrições de Acesso: Propriedades *Somente-Leitura*

- No exemplo anterior, a propriedade `area` é pública.
- Faz sentido: queremos permitir a sua leitura.
- **Mas**: também permite a alteração externa do seu valor.
- Solução: tornar apenas a escrita (`set`) privada: 

````kotlin
val PI = 3.14159
class GrowingCircle(private var radius: Double){
    var area = PI * radius * radius
        private set
    private fun updateArea() {
        area = PI * radius * radius  // Atualiza a área
    }
    fun grow() { 
        radius++
        updateArea()
    }
}
val gc = GrowingCircle(5.0)
println(gc.area)
gc.grow()
println("Nova área: ${gc.area}")
// gc.area = 5.0     // Erro!
````

# *Getters* Personalizados

- No exemplo anterior, toda chamada a `.grow()` dispara um re-cálculo da área.
    - Mesmo que não haja uma consulta logo a seguir.
- Se chamadas a `.grow()` são muito mais frequentes que leituras da área, isto é ineficiente.
- Além disto, suponha que o programa manipule muitos milhões de círculos.
    - Minimizar o gasto de memória pode ser importante.
- Podemos resolver estes problemas transformando `area` numa **propriedade calculada**.
    - Também chamada de ***getters* personalizado**.

# *Getters* Personalizados: Exemplo

````kotlin
val PI = 3.14159
class GrowingCircle(private var radius: Double){
    val area: Double 
        get() = PI * radius * radius 

    fun grow() = radius++
}
val mc = GrowingCircle(5.0)
println("Área do círculo é ${mc.area}")
mc.grow()
println("Área do círculo é ${mc.area}")
````

- **Observações:**
    - O acesso à propriedade é feito normalmente.
    - Relativamente ao exemplo anterior, `area` foi declarada como `val`.
        - Por ser uma propriedade computada, nada é armazenado e, portanto, não faz sentido "alterar" o seu valor.
        - Não há ***backing-field***.

# *Setters* Personalizados

- Podemos personalizar também a escrita à uma propriedade.
- Útil quando a classe usa uma representação interna especial do valor da propriedade.
- Ou quando queremos fazer validações do valor.
- Exemplo: 

````kotlin
class Temperatura {
    var celsius: Double = 0.0
    // Propriedade computada com leitura e escrita
    var fahrenheit: Double
        get() = (celsius * 9 / 5) + 32
        set(value) {
            celsius = (value - 32) * 5 / 9
        }
}
````

# *Getters* Personalizados *vs.* Métodos

- No exemplo anterior, digamos que só estamos interessados em ler a temperatura em Fahrenheit.
- Poderíamos reescrever a solução com um método ao invés de um *getter*:

````kotlin
class Temperatura {
    var celsius: Double = 0.0
    fun fahrenheit(): Double = (celsius * 9 / 5) + 32
}
````

- Quando usar métodos? Quando usar *getters*?
- A recomendações geral é usar *getters* quando:
    - O cálculo é simples, rápido e barato em termos de processamento.
    - O resultado reflete um estado ou característica do objeto.
    - A chamada não tem efeitos colaterais (não altera o estado de outros objetos).

# Funções de Extensão

- Suponha que queremos calcular a raiz quadrada de um valor inteiro.
- Podemos escrever a seguinte função:

````kotlin
fun sqrt(i: Int) = sqrt(i.toDouble()).toInt()
println(sqrt(5))
````

- No contexto do programa, talvez esta operação seja tão comum que talvez fizesse mais sentido que a função fosse, na verdade, um método da classe `Int`.
- Mas não temos acesso ao código de `Int`.
- Podemos contornar isto criando uma **função de extensão**.

````kotlin
fun Int.sqrt() = sqrt(this.toDouble()).toInt()
println(5.sqrt())
````

- Numa função de extensão, a palavra reservada `this` refere-se ao objeto sobre o qual a função é executada.

# Funções Membro *vs.* Funções de Extensão

- Há diferenças importantes entre funções membro (métodos) e funções de extensão.
    - Podem determinar quando usar uma ou outra forma.
- São elas:
    - Definir uma função membro requer acesso à definição da classe.
    - Funções de extensão só têm acesso a **membros públicos** da classe.
    - Funções de extensão são resolvidas de forma **estática em tempo de compilação**. Isto impede a sua utilização em cenários de **polimorfismo**.
        - Mais sobre isto em breve. 

# *Data Classes*

- Classes marcadas com o modificador `data`.

````kotlin
data class Point(val x: Double, val y: Double)
````

- Diz ao compilador para criar alguns métodos especiais.
    - Métodos úteis para **classes que armazenam dados**.
    - `.toString()`, `.equals()`, `.copy()`.

# *Data Classes*: `.toString()`

- Numa *data class*, o `.toString()` gerado automaticamnete imprime:
    - o nome do tipo;
    - os valores de cada propriedade com o respectivo nome.  

````kotlin
data class Point(val x: Double, val y: Double)
val p1 = Point(1.5, 2.0)
println(p1.toString()) // --> Point(x=1.5, y=2.0)
println(p1) // Equivalente: toString() é chamada implicitamente.
````

# *Data Classes*: `.equals()`

- Função `.equals()` é chamda implicitamente quando usamos o operador `==`.
- Numa *data class*, o `.equals()` gerado automaticamnete compara dois objetos propriedade a propriedade:

````kotlin
data class Point(val x: Double, val y: Double)
val p1 = Point(1.5, 2.0)
val p2 = Point(1.5, 2.0)
println("p1==p2: ${p1 == p2}") // --> p1==p2: true
val p3 = Point(2.0, 2.0)
println(p3.equals(p1)) // --> false
````

- Note como o comportamento difere, caso a classe não seja declarada como `data`:

````kotlin
class Point(val x: Double, val y: Double)
val p1 = Point(1.5, 2.0)
val p2 = Point(1.5, 2.0)
println("p1==p2: ${p1 == p2}") // --> p1==p2: false
````

# *Data Classes*: `.copy()`

- Cria uma cópia do objeto.
    - Nova instância do mesmo tipo com os mesmos valores das propriedades.
- Opcionalmente, podemos especificar valores diferentes para algumas das propriedades.

````kotlin
data class Point(val x: Double, val y: Double)
val p1 = Point(1.5, 2.0)
val p4 = p1.copy()
println(p4) // --> Point(x=1.5, y=2.0)
println(p1.copy(y=0.0)) // --> Point(x=1.5, y=0.0)
````
