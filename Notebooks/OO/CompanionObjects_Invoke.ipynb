# *Companion Objects* e Operador *Invoke*

# *Companion Objects*

- É um objeto ***Singleton*** associado a uma **classe**.
    - E não às suas instâncias individuais.
- Útil para definir propriedades e métodos que dizem respeito à classe, mas não a instâncias individuais.
- Exemplo:



```kotlin
class Utilizador(val nome: String) {   
    companion object {
        const val MIN_LENGTH = 3
        
        fun criarAnonimo(): Utilizador {
            return Utilizador("Anónimo")
        }
    }
}
print("Entre nome (min.: ${Utilizador.MIN_LENGTH} letras) -> ")
val username = readln()
val user = if (username.trim().isEmpty()) Utilizador.criarAnonimo() else Utilizador(username)
println(user.nome)
```

# O Operador `invoke`

- Permite "chamar" um objeto como se fosse uma função.

```kotlin
class Saudacao(val prefixo: String) {
    
    operator fun invoke(nome: String) {
        println("$prefixo, $nome!")
    }
}

val saudar = Saudacao("Olá")
saudar.invoke("Maria")
// Chamada usando a sintaxe reduzida (invocação direta do objeto)
saudar("João") // Saída: Olá, João!
```

# O Operador `invoke` e *Companion Objects*

- Um operador `invoke` implementado num *Companion Object* permite "chamar" uma **classe** como se fosse uma função.

```kotlin
class Utilizador(val nome: String) {   
    companion object {
        const val MIN_LENGTH = 3
        operator fun invoke() = "Representa um utilizador com um dado nome"
    }
}
println(Utilizador()) // Equivalente: Utilizador.invoke()
```

- **Nota**:
    - Esta notação (`Utilizador()`) assemelha-se à utilizada para aceder ao construtor da classe (`Utilizador("Afonso")`).
    - Se há um construtor **público** na classe com os mesmos parâmetros do `invoke()`, o construtor é chamado.
