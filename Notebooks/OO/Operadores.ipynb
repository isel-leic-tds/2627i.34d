- Em Kotlin, uma expressão `a+b` é transformada em `a.plus(b)`.
- Funciona se há associada ao tipo de `a` uma função chamada `plus()` com:
    - o modificador `operator`; e
    - Parâmetro do tipo de `b`.

# Operadores `==` e `===`

- Em Kotlin, há dois operadores de igualdade:
    - `==`: igualdade estrutural.
        - Os dois objetos são **iguais** ou **equivalentes**?
    - `===`: igualdade referencial.
        - Os dois objetos são **o mesmo**? 

```kotlin
val l1 = listOf("Ana", "Diogo", "Pedro")
val l2 = listOf("Ana", "Diogo", "Pedro")
println("l1 é igual/equivalente a l2? ${l1 == l2}")
println("l1 é o mesmo objeto que l2? ${l1 === l2}")
```

# Operadores `==` e `!=`

- Similar ao operador `+`, `==` corresponde a uma função denominada `.equals()`
    - `a == b` é compilado para `a?.equals(b) ?: (b === null)`
    - `a != b` é compilado para a negação de `==`.
- Função `.equals()` deve ser:
    - Método ou extensão do tipo de `a`.
    - Do tipo `(Any?) -> Boolean`.
        - Ou seja: deve-se permitir comparação com qualquer tipo anulável.

# Operadores `==` e `!=` (II)

- Toda classe em Kotlin já tem uma implementação por omissão de `.equals()`.
    - Provida pela classe `Any`.

```kotlin
class Circle(val radius: Double)
val c1 = Circle(5.0)
val c2 = Circle(5.0)
val c3 = Circle(10.0)
println(c1 == c3)   // false
println(c1 == c2)   // false
```

- Por que `c1 == c2` é falso?
    - Porque a implementação de `.equals()` de `Any` **testa igualdade referencial**. 

# Operadores `==` e `!=` (III)

- Implementando uma função `.equals()`:

```kotlin
class Circle(val radius: Double) {
    override operator fun equals(other: Any?) = if (other is Circle) other.radius == radius else false
}
val c1 = Circle(5.0)
val c2 = Circle(5.0)
val c3 = Circle(10.0)
println(c1 == c3)   // false
println(c1 == c2)   // true
```

- **Notas**:
    - Modificador `override`: necessário para substituirmos a implementação herdada de `Any`.
    - Operador `is`: testa o tipo de um objeto.

# A Função `.compareTo()`

- **Realiza comparações relativas**.
    - Anotada com o modificador `operator`.
    - Mapeada a partir dos operadores `<`, `>`, `<=`, `>=` .
    - Retorna um `Int`:
        - `0` se forem iguais.
        - Negativo se `this` for menor que o parâmetro.
        - Positivo se `this` for maior que o parâmetro.
- **Transformação do Compilador**:
  - `a > b`  → `a.compareTo(b) > 0`
  - `a < b`  → `a.compareTo(b) < 0`
  - `a <= b` → `a.compareTo(b) <= 0`
  - `a >= b` → `a.compareTo(b) >= 0`

# A Função `.compareTo()`: Exemplo de Implementação

- Comparar dois círculos.

```kotlin
class Circle(val radius: Double) {
    /*...*/
    operator fun compareTo(other: Circle) = 
        if (this.radius < other.radius) -1 
        else if (this.radius > other.radius) 1 
        else 0
}
val c1 = Circle(5.0)
val c2 = Circle(5.0)
val c3 = Circle(10.0)
println(c1 < c2)   // false
println(c1 <= c2)   // true
println(c3 >= c2)   // true
```

- **Notas**:
    - `Any` não implementa `.compareTo()`.
        - Logo, não há `override`.
    - Diferente da `.equals()`, `.compareTo()` tem parâmetro do mesmo tipo da classe.

# A Função `.toString()`

- **Propósito**: retorna representação textual do objeto.
- Classe `Any` provê implementação por omissão.
- **Mas**:
    - Retorna apenas uma identificação interna do objeto.
    - `NomeDaClasse@HashCodeHexadecimal` (ex: `Date@b7c0b`)

```kotlin
class Circle(val radius: Double) {
    /*...*/
}
val c1 = Circle(5.0)
println(c1.toString())
```

# A Função `.toString()`: Exemplo de Implementação

```kotlin
class Circle(val radius: Double) {
    override fun toString() = "I'm a circle of radius %.02f".format(radius)
}
val c1 = Circle(5.0)
println(c1.toString())
```

- **Notas**:
    - `override` para substituir a implementação por omissão da `Any`.
    - Função `.format()`: formatações especiais de dados para *string*.
        - *e.g.*, `%.02f`:
            - Dado é um número real (`f`).
            - Usar dois algarismos depois do separador decimal (`.02`)
