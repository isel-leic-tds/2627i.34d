# *Selead Classes* e *Pattern Matching*

# *Pattern Matching*

- **Padrão de projeto** que permite casar ações com tipos ou estrutura de dados.
- Em Kotlin, normalmente é implementada com base em dois elementos:
    - Uma hierarquia de classes **fechada**.
    - Uma estrutura de decisão do tipo `when`. 

# Hierarquia de Classes Fechada

- Uma hierarquia de classes é fechada quando a classe base tem o modificador `sealed`.
- Uma classe `selead` implica:
    - Só pode ser **estendida por outras classes** que:
        - São conhecidas no momento da compilação
        - Estão **no mesmo pacote e módulo**.
    - É uma classe **abstrata**.

# Hierarquia de Classes Fechada: Exemplo

````kotlin
sealed class Result

data class Success(val data: String) : Result()
data class Error(val exception: Throwable) : Result()
object Loading : Result()
````

- Fora do módulo que declara a classe `Result`, não é possível estendê-la.
- A classe `Result` é também implicitamente abstrata: não pode ser instanciada.
- Geralmente utilizado para representar **estados** dentro de um sistema.

# *Pattern Matching*: Exemplo

```Kotlin
fun processarResultado(result: Result): String = when (result) {
    is Success -> "Dados recebidos: ${result.data}"
    is Error -> "Ocorreu um erro: ${result.exception.message}"
    Loading -> "Carregando..."
}
```

- Estrutura `when` usada para processar `Result` de forma **diferente a depender do tipo específico**.
- Note que `when` é exaustivo, porque classe `Result` é `sealed`.
    - Só há as três possibilidades já declaradas: `Success`, `Error` e `Loading`.
    - Todas as possibilidades são cobertas.
        - `Result` ser  `sealed` permite ao compilador verificar isto.
