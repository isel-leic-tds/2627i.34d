# Coleções Genéricas

- Coleções já estudadas até aqui são **genéricas**.
    - *e.g.*, mesma classe `List` serve para listas de `Int`, `Double`, `String`, ...
    - *e.g.*, mesma classe `MutableStack` serve para *stacks* de `Int`, `Double`, `String`, ...
- Essencialmente, estas classes podem ser **parametrizadas** por um tipo:

````kotlin
val l1: List<String> = listOf<String>("Ana", "Beatriz", "Carlos")
val l2: List<Int> = listOf<Int>(1, 2, 3, 4)
````

# Declaração de uma Classe Genérica

- Podemos declarar uma classe parametrizada por um tipo `T` com a notação `<T>`:

````kotlin
class Caixa<T>(var conteudo: T) {
    fun obter(): T {
        return conteudo
    }

    fun atualizar(novoConteudo: T) {
        conteudo = novoConteudo
    }
}
val caixaString1 = Caixa<String>("teste 1")
val caixaString2 = Caixa("teste 2")
val caixaInt = Caixa(3)
println(caixaString2.obter())
````

- **Notas**:
    - Tipo `T` pode ser usado dentro da classe em qualquer contexto que um nome de tipo seria.
        - Tipo de parâmetro, tipo de retorno de função, tipo de valor, ... 

# Múltiplos Parâmetros de Tipo

- É possível ter mais de um parâmetro de tipo:

````kotlin
class Par<T, K>(val chave: T, val valor: K)

val p1 = Par<String, Int>("Maria", 3)
val p2 = Par("João", 1)
````

# Funções Genéricas 

- Funções também podem ser parametrizadas quanto ao tipo.
- Exemplo:

````kotlin
fun <T> logObject(obj: T) = println("LOG: ${obj.toString()}")

logObject("Pedro")
logObject(3.1415)
````

# Parâmetros de Tipo e Funções de Extensão

- Funções de extensão podem fazer referência aos parâmetros de tipo de uma classe.
- Exemplo:

````kotlin
class Par<T, K>(val chave: T, val valor: K)

fun <T, K> Par<T, K>.getChave(): T = this.chave
fun <T, K> Par<T, K>.getValor(): K = this.valor

val p2 = Par("João", 1) 
println("chave: ${p2.getChave()}; valor: ${p2.getValor()}")
````

# *Type Erasure*

- Código compilado perde informação sobre tipos genéricos.
    - Denominado *type erasure*. 
- Exemplo:

```Kotlin
fun verificarLista(list: Any) {
    // Erro de compilação: Cannot check for instance of erased type
    if (list is List<String>) { 
        println("É uma lista de Strings")
    }
}
```

- Em tempo de execução, não é possível saber se um objeto é `List<String>` ou `List<Int>`

# *Type Erasure*: *Star-Projection*

- Permite denotar um objeto de uma determinada classe, independentemente do seu tipo genérico.

````kotlin
val list1 = listOf("Maria", "José", "Diogo")
val list2 = listOf(5, 4, 3, 2, 1)
if (list1 is List<*>) println("list1 é uma lista de algum tipo")
if (list2 is List<*>) println("list2 é uma lista de algum tipo")
````

- Com múltiplos parâmetros de tipo:

````kotlin
class Par<T, K>(val chave: T, val valor: K)

val p1 = Par<String, Int>("Maria", 3)

if (p1 is Par<*, *>) println("p1 é um par de algum tipo")
````
