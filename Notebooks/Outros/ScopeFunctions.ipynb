# *Scope Functions*

- Permitem executar blocos de código no contexto de um objeto.
- Recebem como parâmetro uma lambda executada no escopo do objeto.
    - Na lambda, objeto corresponde ao `it` ou ao `this`, a depender da *scope function*.
- Diversas variantes. Algumas comuns:

| **Função** | **Referência do Objeto** | **Valor de Retorno** | **Caso de Uso Principal**                                                           |
|------------|--------------------------|----------------------|-------------------------------------------------------------------------------------|
| let        | it                       | Resultado da lambda  | Operações em objetos não-nulos (?.let) e transformação de resultados.               |
| apply      | this                     | O próprio objeto     | Configuração e inicialização de objetos.                                            |
| also       | it                       | O próprio objeto     | Ações adicionais/colaterais que não alteram o objeto (ex: logs, remoção e retorno). |
| with       | this                     | Resultado da lambda  | Agrupar chamadas de métodos em um objeto (usado como with(obj) { ... }).            |

```kotlin
// Assinatura e funcionamento da função de scope inline 'also'
public inline fun <T> T.also(block: (T) -> Unit): T {
    block(this)
    return this
}
```

# *Scope Functions*: `.let()`

- Usada em conjunto com o *safe-call operator*: `?.`
- Bloco só é executado se objeto não é `null`.

```kotlin
//val nome: String? = "Kotlin"
val nome: String? = null

nome?.let {
    // 'it' refere-se à variável 'nome' não-nula
    println("O tamanho do nome é ${it.length}")
}
```

- **Notas**:
    - Dentro da lambda, `it` certamente não é `null`.
    - *Smart cast*: `it` é do tipo `String`, e não `String?`. 

# *Scope Functions*: `.apply()`

- Permite aplicar configurações sobre um objeto.
    - *e.g.*, aplicar valores às propriedades.
- Retorna o objeto.

```kotlin
class Pessoa {
    var nome: String = ""
    var idade: Int = 0
}

val pessoa = Pessoa().apply {
    nome = "Ana"  // equivalente a this.nome = "Ana"
    idade = 30
}
println(pessoa.nome)
println(pessoa.idade)
```

# *Scope Functions*: `.also()`

- Permite executar ações que não alteram o objeto.
- Retorna o objeto.

```kotlin
class Pessoa (val nome: String = "", val idade: Int = 0)
val pessoa = Pessoa("Gonçalo", 25).also {
    println("Criação de pessoa chamada ${it.nome}")
}
```

# *Scope Functions*: `.with()`

- Permite fazer várias chamadas a métodos do objeto, sem repetir o seu nome.
    - Recebe objeto como parâmetro. 
- Retorna o resultado da lambda.


```kotlin
class Pessoa (val nome: String = "", var idade: Int = 0) {

    fun adicionaAnos(anos: Int = 1): Int {        
        idade += anos
        return idade
    }
}

val informacao = with(Pessoa("João", 45)) {
    adicionaAnos()
    "Nome: $nome, Nova Idade: $idade"
}
println(informacao)
```
