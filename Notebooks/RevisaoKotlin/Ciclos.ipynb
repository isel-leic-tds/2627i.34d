# Revisão de Kotlin: Ciclos

# Ciclos

- **Ciclos definidos**: `for`.
    - Corre para os elementos específicos de uma sequência.
- **Ciclos indefinidos**: `while` e `do...while`.
    - Correm enquanto certa condição é verdadeira.   

# Ciclo `for`

- **Sintaxe**:
    - Palavra reservada `for`
    - Especificação da sequência entre parêntesis.
        - Variável de iteração.
        - Operador `in`.
        - Objeto iterável (*e.g.*, sequência, lista, ...)
    - Comando ou bloco delimitado por chavetas.

```kotlin
for(i in 0 ..< 5) print("$i ")            // --> 0 1 2 3 4
for(n in listOf("Pedro","Ana","Maria")) print("$n ") // --> Pedro Ana Maria
for(letter in "Kotlin") print("$letter ")  // --> K o t l i n
```

# Ciclo `while`

- **Sintaxe**:
    - Palavra reservada `while`
    - Condição entre parêntesis.
        - Qualquer valor *booleano*.
    - Comando ou bloco delimitado por chavetas.

```kotlin
var remaining = 3
while(remaining > 0) {
   print("Falta $remaining ") // --> Falta 3 Falta 2 Falta 1
   remaining--   // Equivalente a: remaining = remaining - 1 ou remaining -= 1
} 
while (remaining > 0)
   println("Acabou.") // Nunca será executado
```

# Ciclo `do...while`

- Similar ao `while`, mas condição é avaliada **após** cada iteração.

```kotlin
var count = 0
do {
   count++
   print("$count ")
} while (count <= 3) // --> 1 2 3 4

do println("Ok") while (count < 1) // --> Ok
```

# Instrução `continue`

- Interrompe a iteração atual, pulando para a próxima
- Aplicável a qualquer tipo de ciclo.

```kotlin
for (i in 1..10) {
   print(i)
   if (i % 2 == 0) {
       println()
       continue // Ignorar números pares
   }
   println(" é ímpar")
}
```

# Instrução `break`

- Interrompe a iteração atual **e o restante do ciclo**.
- Aplicável a qualquer tipo de ciclo.

```kotlin
for (i in 5..10) {
   if (i % 4 == 0) break  // Break the loop
   println("Número: $i")
}
```
