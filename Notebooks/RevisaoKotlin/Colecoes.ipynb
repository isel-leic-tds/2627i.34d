# Revisão de Kotlin: Coleções

# Coleções

* Estruturas para armazenar múltiplos elementos. Dividem-se em variações imutáveis (somente leitura) e mutáveis.

| Tipo | Descrição | Repetições |
| ----- | :---: | :---: |
| **List\<T\>** | Sequência ordenada de elementos (T) | Pode ter repetições |
| **Set\<T\>** | Conjunto não ordenado de elementos (T) | Sem repetições |
| **Map\<K,T\>** | Conjunto de pares Chave (K) \- Valor (T) | Sem chaves repetidas |

# Listas

* Coleção **ordenada** com elementos acessíveis por índice.
* Criada no formato imutável com `listOf()`:

```kotlin
val names: List<String> = listOf("Pedro", "Ana", "João", "Maria", "Ana")
println(names)            // --> [Pedro, Ana, João, Maria, Ana]
```

* Sintaxe experimental com colchetes: `val names: List<String> = ["Pedro", "Ana"]`.
* Acceso por índice via `[index]`:

```kotlin
println("First name: ${names[0]}")            // --> First name: Pedro
println("Last name: ${names[names.size-1]}")  // --> Last name: Ana
```

# Listas: `first()` e `last()`

* Funções `.first()` e `.last()` obtêm as extremidades da lista:

```kotlin
println("First name: ${names.first()}")  // --> First name: Pedro
println("Last name: ${names.last()}")    // --> Last name: Ana
```

# Listas: `in`

* Verificação de pertencimento via operador `in`:

```kotlin
println("Maria" in names)    // --> true
```

# Listas: Concatenação e Remoção

* Concatenação com operador `+` (gera uma nova lista imutável):

```kotlin
val notas = listOf(15, 15, 11, 17)
val novasNotas = notas + listOf(20, 19)
println(novasNotas)
```

* Operador `-` remove a primeira ocorrência do elemento (retorna uma nova lista):

```kotlin
val notas = listOf(15, 15, 11, 17)
println(notas)   // [15, 15, 11, 17]
val novasNotas = notas - 15
println(novasNotas) // [15, 11, 17]
println(notas)   // [15, 15, 11, 17]
```

# Listas Mutáveis

* **`MutableList`**: Criada com `mutableListOf()`. Suporta métodos de alteração direta (`add()`, `remove()`, indexação para substituição):

```kotlin
val mutableNames: MutableList<String> = mutableListOf("Ana", "Pedro", "Maria")
mutableNames.add("Ana")        // Append at the end
mutableNames.remove("Pedro")   // Remove the first found
mutableNames[1] = "Isabel"     // Replace the element at index 1
println(mutableNames)          // --> [Ana, Isabel, Ana]
```

* Listas mutáveis podem ser convertidas implicitamente (casting) em imutáveis.

```kotlin
val readOnlyNames: List<String> = mutableNames
```

# Conjutos

- Armazenam elementos **sem ordem** e **sem repetições**.
- Úteis para avaliação de **pertinência**.
    - Consultas mais rápidas que em listas.
- **Versão imutável**: `Set`
    - Instanciado com a função `setOf()`:

```kotlin
val fruit: Set<String> = setOf("maçã", "banana", "kiwi", "laranja", "kiwi")
println(fruit)           // --> [maçã, banana, kiwi, laranja]
```

# Conjuntos: `.size`, `.first()` e `.last()`

- **Propriedade** `.size`: número de elementos no conjunto.
- **Funções** `.first()` e `.last()`: análogas às dos intervalos.

```kotlin
println("|fruit| = ${fruit.size}")            // --> |fruit| = 4
println("${fruit.first()}, ${fruit.last()}")  // --> maçã, laranja
```

- **Observação**: elementos não têm ordem, então não faz sentido indexar um `Set`.

# Conjuntos: Pertinência, Adições e Remoções

- Operador `in`: verifica se certo elemento pertence ao `Set`:

```kotlin
println("banana" in fruit)    // --> true
```

- Operadores `+` e `-` adicionam e removem elementos.
    - Mas `Set` é imutável, então **resultam num novo** `Set`: 

```kotlin
var moreFruit = fruit + setOf("pera", "banana") // A new Set<String> is created
moreFruit = moreFruit - "kiwi"             // Set<String> - String = Set<String>
println(moreFruit)                    // --> [maçã, banana, laranja, pera]
```

# Conjuntos Mutáveis

- Tipo `MutableSet`.
    - Instanciado com a função `mutableSetOf()`.
    - Provê funções `.add()` e `.remove()`.

```kotlin
val mutableFruit: MutableSet<String> = mutableSetOf("maçã", "banana", "kiwi")
println(mutableFruit)          // --> [maçã, banana, kiwi]
mutableFruit.add("morango")    // Append at the end
mutableFruit.remove("banana")  // Remove
println(mutableFruit)          // --> [maçã, kiwi, morango]
```

- Como outras coleções, versão mutável pode ser convertida implicitamente para a imutável:

```kotlin
val readOnlyFruit: Set<String> = mutableFruit
```

# Mapas

- Armazenam pares `(chave, valor)`.
    - Pense numa lista, mas indexada por uma chave que não precisa ser sequencial nem numérica.
- **Versão imutável**: tipo `Map`.
    - Instanciada com a função `mapOf()`.
    - **Mas**: elementos são pares (`Pair`).
        - Criados com `Pair()` ou operador `to`.     

```kotlin
val p1= Pair("Pedro",48)
val personAges: Map<String,Int> = mapOf(p1, "Ana" to 46, "Luis" to 27, "João" to 15)
println(personAges)        // --> {Pedro=48, Ana=46, Luis=27, João=15}
```

# Mapas: Indexação e Tamanho

- Acesso aos elementos é similar a uma lista: feita por indexação.
    - `[chave]`
- Como outras coleções, há uma propriedade `.size`.

```kotlin
println(personAges.size)      // --> 4
println(personAges["Luis"])   // --> 27
```

# Mapas: Pesquisa

- Como em outras coleções, há o operador `in`.
- **Importante**: pesquisa se a **chave existe** no mapa:

```kotlin
println("Ana" in personAges)    // --> true
```

- Mas:

```kotlin
println(46 in personAges)    // erro de compilação!
```

# Mapas: Propriedades `.keys` e `.values`

- Retornam coleções dos elementos armazenados no mapa.
    - `.keys`: todas as chaves.
    - `.values`: todos os valores.

```kotlin
val keys: Set<String> = personAges.keys
println(keys)               // --> [Pedro, Ana, Luis, João]
println(personAges.values)  // --> [48, 46, 27, 15]
```

# Mapas Mutáveis

- Análogos às versões mutáveis das demais coleções.
- Diferença notável:
    - Adição é feita pela função `.put()`. 

```kotlin
val mutablePersonAges = mutableMapOf("Pedro" to 48, "Ana" to 46, "Luis" to 27)
mutablePersonAges.put("João", 15)
mutablePersonAges.remove("Ana")
mutablePersonAges["Pedro"] = 49
println(mutablePersonAges) // --> {Pedro=49, Luis=27, João=15}
val imutablePersonAges: Map<String,Int> = mutablePersonAges
println(imutablePersonAges)
```
