# Revisão de Kotlin: Estruturas de Decisão

# Estruturas de Decisão

* Kotlin possui duas estruturas principais: `if` (binária) e `when` (múltipla).

### Decisão Binária

* Sintaxe: `if (condição) { ... } else { ... }`.
* Blocos com apenas uma linha não necessitam obrigatoriamente de `{...}`.

````kotlin
val value: Int = (-10..10).random()
if (value >= 0) {
   println("$value is positive.")
   println("So $value can be 3.")
}
else {
   println("$value is negative.")
   println("So $value cannot be 3.")
}

var finalValue = 0
if (value == 3)
   finalValue = -value
else
   finalValue = value

println(finalValue)
````

# Estruturas de Decisão: `if` como expressão

* **`if` como expressão**: Pode retornar um valor diretamente, substituindo o operador ternário.

````kotlin
val value: Int = (-10..10).random()
val finalValue = if (value == 3) -value else value

println(finalValue)
````

* Quando usado em blocos, o `if` retorna o valor da **última linha** do bloco executado:

````kotlin
val value: Int = (-10..10).random()
val finalValue = if (value >= 0) {
   println("$value is positive.")
   println("So $value can be 3.")
   5
}
else {
   println("$value is negative.")
   println("So $value cannot be 3.")
   1
}

println(finalValue)  // 5 ou 1 a depender da condição do if
````

* **Nota**: Quando usado como expressão, o bloco `else` é obrigatório.

# Estruturas de Decisão: Decisão Múltipla

* A estrutura `when` é utilizada para múltiplos fluxos de execução baseados em condições:

````kotlin
val value: Int = readln().toInt()
when {
   value !in -10..10 -> {
       println("Error.")
       println("$value is invalid.")
   }
   value > 0 -> println("$value is positive.")
   value < 0 -> println("$value is negative.")
}
````

# Estruturas de Decisão: `when` com Sujeito

* **`when` com sujeito**: Para comparar diretamente uma variável com múltiplos valores ou condições:

````kotlin
val value: Int = readln().toInt()
when (value) {
   !in 1..10 -> {
       println("Error.")
       println("$value is invalid.")
   }
   2, 3, 5, 7 -> println("$value is prime.")
}
````

* Permite combinar múltiplos valores separados por vírgula (`2, 3, 5, 7`).

# Estruturas de Decisão: `else`

* Use a cláusula `else` para tratar casos não explicitamente avaliados:

````kotlin
val value: Int = readln().toInt()
when (value) {
   !in 1..10 -> {
       println("Error.")
       println("$value is invalid.")
   }
   2, 3, 5, 7 -> println("$value is prime.")
   else -> println("$value is not prime.")
}
````

# Estruturas de Decisão: `when` como Expressão

* **`when` como expressão**: Retorna o valor do ramo correspondente (requer ser exaustivo):

````kotlin
val value = (-10..10).random()
val result = when {
   value > 0 -> "positive"
   value < 0 -> "negative"
   else -> "zero"
}
println("$value is $result.")
````

# Estruturas de Decisão: Expressões de Guarda

* **Condições de guarda (`if`)**: Suportadas a partir do Kotlin 2.1 dentro das hipóteses do `when`:

```kotlin
val month = 2
val year = 2026
val days = when(month) {
   4, 6, 9, 11 -> 30
   2 if (year % 4 == 0 && (year % 100 != 0 || year % 400 == 0)) -> 29
   2 -> 28
   else -> 31
}
println("O mês $month do ano $year tem$days dias.")
```
