# Revisão de Kotlin: Funções

# Funções

- Executam "tarefas" dentro de um programa.
    - Cada função deve ter **uma única responsabilidade** bem definida.
- Podem ou não retornar valores.
- Funções podem ser **nomeadas** ou **anónimas/lambdas**.

# Funções Nomeadas: Declaração

- Funções nomeadas são declaradas com a palavra chave `fun`: 

```kotlin
fun readAge(name: String, default: Int): Int {
   print("How old is $name? ")      // Displays the question with the name.
   val input = readln().trim()      // Reads the input ignoring the spaces.
   if (input.length==0 || input[0] !in '0'..'9') // If no input or invalid.
       return default               // Returns the default value.
   return input.toInt()             // Converts the input to Int and returns it.
}
```

- Sintaxe:
    - `fun`
    - Nome da função.
    - Lista de parâmetros entre parêntesis (com respetivos tipos).
    - [opcional] Tipo do retorno.
    - Corpo da função entre chavetas.
        - Se função retorna algo, deve conter um `return`. 

# Funções Nomeadas: Chamada

- Indica-se o nome e os parâmetros entre parêntesis.
    - Mesmo que não haja parâmetros, **parêntesis são obrigatórios**.

```kotlin
val person = "Maria"
val age = readAge(person, 18)  // Call the function and save the value.
println("$person are $age years old.")
```

# Parâmetros Nomeados

- Ao chamar uma função, **a ordem dos argumentos importa**.
    - Devem ser passados na ordem da declaração da função.
- Alternativa: especificar os nomes dos parâmetros.
    - Melhora a **legibilidade do código**.
    - Permite alterar a ordem dos argumentos. 

```kotlin
//val age = readAge(18, person)                // Erro porque argumentos estão fora de ordem.
val age1 = readAge(person, default = 18)       // Ordem correta.
val age2 = readAge(default = 18, name = person) // Ordem errada, mas argumentos são nomeados.
```

# Parâmetros com Valor por Omissão

- Ao declarar uma função, certos parâmetros podem ter valores por omissão.
    - Chamador não precisa especificá-los, se não quiser.  

```kotlin
fun readAge(name: String, default: Int = 18): Int {
   print("How old is $name? ")      // Displays the question with the name.
   val input = readln().trim()      // Reads the input ignoring the spaces.
   if (input.length==0 || input[0] !in '0'..'9') // If no input or invalid.
       return default               // Returns the default value.
   return input.toInt()             // Converts the input to Int and returns it.
}

println(readAge("Maria"))
```

# Tipo de Retorno `Unit`

- Uma função que não precisa de retornar nada, retorna `Unit`:

```kotlin
fun main(): Unit {
  //...
}
```

- Neste caso, a especificação do tipo de retorno é opcional:

```kotlin
fun main() {   // Equivalente ao anterior
  //...
}
```

- Funções que não retornam nada não precisam especificar explicitamente um `return`.
    - Mas **podem** fazê-lo para interromper a execução imediatamente.
        - Análogo ao `break` para ciclos.

```kotlin
fun hello(name: String = "") {
   if (name == "") {
       println("Hello, World!")
       return
   }
   println("Hello, $name!")
}
```

# Funções Expressão

- Às vezes, funções apenas computam e retornam o valor de uma expressão simples.
- Kotlin fornece uma notação mais compacta para estes casos.
- Suponha uma função que calcula o máximo de dois valores:

```kotlin
fun max(a: Int, b: Int): Int {
   return if (a >= b) a else b
}
```

- Pode ser reescrita como:

```kotlin
fun max(a: Int, b: Int): Int =
   if (a >= b) a else b
```

- Adicionalmente, o compilador consegue inferir o tipo de retorno. Logo, de forma ainda mais compacta:

```kotlin
fun max(a: Int, b: Int) = if (a >= b) a else b
```

# Funções como Valores de Primeira Classe

- Em Kotlin, funções são **valores de primeira classe**.
    - *i.e.*, podem ser usadas em qualquer contexto em que valores quaisquer seriam usados.
- Exemplos:
    - Podem ser passadas como **parâmetros para outras funções**.
    - Podem ser **armazenadas em variáveis**.
    - Podem ser **retornadas por outras funções**.
    - ...

```kotlin
fun odd(n: Int) = (n % 2 != 0)
val minhaFuncao = ::odd  // O operador :: retorna uma referência para a função
println(minhaFuncao(3))  // Chama a função armazenada em minhaFuncao
```

# O Tipo de uma Função

- Assim como podemos definir que uma variável ou parâmetro é do tipo `Int` ou `String`, podemos descrever em Kotlin tipos associados a funções.
- O tipo de uma função é determinado por:
    - Quantidade, ordem e tipos dos seus parâmetros.
    - Tipo de retorno.
- Exemplo: o tipo da função `odd()` definada acima é `(Int)->Boolean`
- Exemplo de uso:

```kotlin
fun odd(n: Int) = (n % 2 != 0)
val minhaFuncao: (Int)->Boolean = ::odd  // Declaração explícita do tipo
println(minhaFuncao(3))  // Chama a função armazenada em minhaFuncao
```

# Funções como Parâmetro

- Como podemos declarar uma função que recebe outra como parâmetro?

```kotlin
// Funções que convertem um Char para maiúscula ou minúscula
fun upperOf(c: Char): Char = if (c in 'a'..'z') c - ('a'-'A') else c
fun lowerOf(c: Char): Char = if (c in 'A'..'Z') c + ('a'-'A') else c
// Função que imprime string transformada através de uma função de mapeamento
fun printlnIn(text: String, charMapper: (Char)->Char) {
   for (c in text)
       print(charMapper(c))   // Chamar a função de mapeamento recebida
   println()
}
// Exemplos de uso
printlnIn("Hello, World!", ::upperOf) // --> HELLO, WORLD!
printlnIn("Hello, World!", ::lowerOf) // --> hello, world!
```

# Funções como Valor de Retorno

- Basta definir o tipo da função retornada como tipo de retorno da função:

```kotlin
fun charMapper(upper: Boolean): (Char)->Char = 
    if (upper) ::upperOf 
    else ::lowerOf

printlnIn("Hello, World!", charMapper(upper = true)) // --> HELLO, WORLD!
```

# Expressões Lambda

- Definir uma função nomeada apenas para passá-la como parâmetro é inconveniente.
- Alternativa: **funções anónimas ou expressões lambda**.
    - Funções definidas como expressões, sem nome.
- Sintaxe: `{parâmetros -> expressão}`
- Exemplo:

```kotlin
printlnIn(
    "Hello, World!", 
    { c -> if (c in 'A'..'Z') c + ('a'-'A') else c }
) // --> hello, world!
```

# Expressões Lambda: Notação Alternativa

- Quando a lambda é o último argumento da chamada da função, ela pode ser omitida da lista de parâmetros e colocada após os parêntesis:

```kotlin
printlnIn("Hello, World!") { 
    c -> if (c in 'A'..'Z') c + ('a'-'A') else c 
} // --> hello, world!
```

# Expressões Lambda: `it`

- Quando uma lambda tem apenas um parâmetro, podemos omitir a sua declaração.
    - Passa a haver um parâmetro implícito nomeado `it`.  

```kotlin
printlnIn("Hello, World!") { 
    if (it in 'A'..'Z') it + ('a'-'A') else it 
} // --> hello, world!
```

# Expressões Lambda: Exemplo da Função `.filter()`

- A função `.filter()` é definida sobre coleções.
    - Retorna uma versão "filtrada" da coleção.
        - Contém apenas elementos que atendem a certo critério.
        - Critério é definido por função passada como parâmetro. 

```kotlin
val grades = listOf(7, 13, 5, 18, 3, 14, 11, 15)
val passing = grades.filter { it >= 10 }
println(passing) // --> [13, 18, 14, 11, 15]
```

- A lambda `{ it >= 10 }` retorna `true` apenas para valores maiores que 10.
