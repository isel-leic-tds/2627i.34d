# Revisão de Kotlin: Intervalos

# Intervalos (Ranges) em Kotlin

* Representam sequências finitas de valores com limite inicial e final.
* Usados em laços `for`, verificações `in` / `!in` e coleções.

#### Intervalo Fechado: Operador `..`

* **Operador `..`**: Inclui ambos os limites (inicial e final).

```kotlin
val intervaloFechado = 1..5 
// Contém: 1, 2, 3, 4, 5

println(5 in intervaloFechado) // Imprime: true
```

# Intervalos Abertos à Direita: Operadores `until` e `..<`

* **`until` / `..<`**: Exclui o limite superior (ideal para índices de coleções).

```kotlin
val intervaloAberto = 1 until 5 
// Contém: 1, 2, 3, 4 (o 5 fica de fora)

println(5 in intervaloAberto)
println(4 in intervaloAberto)
```

* Equivalente utilizando o operador `..<`:

```kotlin
val intervaloAberto = 1..<5 
// Contém: 1, 2, 3, 4 (o 5 fica de fora)

println(5 in intervaloAberto)
println(4 in intervaloAberto)
```

# Intervalos: Direção

* **`downTo`**: Para contagens decrescentes.
* Propriedades `.first` e `.last` obtêm as extremidades do intervalo:

```kotlin
val intervalo = 5 downTo 1 
// Contém: 5, 4, 3, 2, 1

println(intervalo.first)
println(intervalo.last)
```

# Intervalos: Controlo de Passo

* **`step`**: Define o incremento/passo.
* **`.count()`**: Retorna a quantidade de elementos no intervalo.

```kotlin
val intervalo = 5 downTo 1 step 2
// Contém: 5, 3, 1

println(intervalo.first)
println(intervalo.last)
println(intervalo.count())
```

# Intervalos com Outros Tipos de Dados

* Funciona com qualquer tipo que implemente `Comparable` (`Char`, `String`, `Double`, etc.):

```kotlin
val letras = 'a'..'e' 

println('c' in letras)
println('z' in letras)
```

* Tipos não-enumeráveis (`Double`, `String`) suportam a criação de limites para verificação:

```kotlin
val prices = 3.5..6.75     // Double range
println(4.5 in prices)     // --> true
val names = "ana".."pedro" // String range
println("joão" in names)   // --> true
println("rui" in names)    // --> false
```

# Intervalos: random

* **`.random()`**: Gera um valor aleatório contido no intervalo.

```kotlin
val dice = (1..6).random() // Random number in the range
println("Dice = $dice")    // --> Dice = ...
```

# Intervalos: Resumo dos Operadores e Funções

| Sintaxe | Descrição | Exemplo |
| --- | --- | --- |
| `a..b` | Inclui `a` e `b` | `1..5` $\rightarrow$ $1, 2, 3, 4, 5$ |
| `a until b` | Inclui `a`, **exclui `b`** | `1 until 5` $\rightarrow$ $1, 2, 3, 4$ |
| `a downTo b` | Contagem regressiva | `5 downTo 1` $\rightarrow$ $5, 4, 3, 2, 1$ |
| `step n` | Define o incremento | `1..5 step 2` $\rightarrow$ $1, 3, 5$ |
| `x in range` | Verifica pertencimento | `3 in 1..5` $\rightarrow$ `true` |
