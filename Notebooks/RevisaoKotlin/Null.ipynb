# Revisão de Kotlin: Null

# O Valor `null`

- `null` representa **ausência de informação**.
    - Exemplo: qual é o valor associado a uma chave inexistente num `Map`?

```kotlin
val personAges: Map<String,Int> = mapOf("Ana" to 46, "Luis" to 27, "João" to 15)
println(personAges["Luis"])   // --> 27
println(personAges["Pedro"])  // --> null
```

- Frequentemente útil, mas potencialmente problemático.
    - E se tentarmos acessar uma propriedade do `null`?

# Tipos *nullable*

- Em geral, tipos em Kotlin não admitem o valor `null`.

```kotlin
var i = 30    // Tipo inferido: Int
// i = null   // Erro de compilação! Int não admite null
```

- Para admitir o valor `null`, a variável ou valor deve ser de um tipo *nullable*.
    - Indicado utilizando o sufixo `?` ao final do tipo.

```kotlin
var i: Int? = 30
i = null    // Válido.
println(i)
```

# Segurança no Uso de `null`

- Certas operações não pode ser realizadas sobre valores `null`.
    - *e.g.*, executar métodos, aceder a propriedades, ...
- Logo, se um valor, variável ou propriedade é de um tipo *nullable*, certas operações falham se não houver uma verificação.

val i: Int? = null
// val s = i + 3   // Erro porque i é de tipo nullable
val j: Int? = 1
// val s = j + 3   // Erro, mesmo que saibamos que j não é null aqui.

- Exemplo de verificação:

```kotlin
val j: Int? = 1
val s = if (j != null) j + 3 else 3   // Válido, porque o compilador sabe (após o teste) que j não é null
println(s)
```

- *Smart cast*: após o `if`, compilador considera tipo de `j` como sendo `Int`, e não `Int?`.

# O Operador Elvis

- No exemplo anterior, fizemos:

```kotlin
val s = if (j != null) j + 3 else 3
```

- Semântica: se `j` é `null`, trate-o como 0; do contrário, use o seu valor.
- Kotlin tem uma notação mais compacta para isto através do **operador Elvis**: `?:`

```kotlin
val j : Int? = null
val s = (j ?: 0) + 3
println(s)
```

- Pode ser usado também nos valores por omissão dos parâmetros:

```kotlin
fun printText(maybeText: String?) = println(maybeText ?: "")

printText("abc")    // --> abc
printText(null)     // --> (empty line)
```

# Acesso Seguro a Propriedades ou Funções

- *Safe-call operator*: `?.`
- Se o valor, variável ou parâmetro é `null`, tentativa de acesso retorna `null`.
    - Caso contrário, acesso normal é realizado. 

```kotlin
val i : Int? = 3
//val d: Double? = i.toDouble()   // Erro: i pode ser null
val d: Double? = if (i != null) i.toDouble() else null    // Válido, mas verboso
val d2: Double? = i?.toDouble()      // Equivalente com o safe-call operator.
```

# Operador !! (Double Bang)

- Em alguns casos, o programador **sabe** que uma variável de tipo *nullable* não tem valor `null`.
    - Mesmo sem um teste explícito.
- Nestes casos, o operador `!!` pode ser utilizado para transformar o tipo *nullable* no tipo *non-nullable* correspondente:  

```kotlin
val personAges: Map<String,Int> = mapOf("Ana" to 46, "Luis" to 27, "João" to 15)
val ageOrNull: Int? = personAges["Luis"] // Supostamente, personAges["Luis"] pode ser null
val age = ageOrNull!!  // !! transforma Int? num Int
println(age::class.simpleName)
```

# Operador !! (Double Bang) (II)

- Mas e se o programador comete um erro?
- E se o operador `!!` é aplicado a um valor que, de facto, é `null`?
- **Resposta**: exceção do tipo NullPointerException (NPE) em tempo de execução.
- Alternativa melhor: utilizar a função `checkNotNull`.

```kotlin
val personAges: Map<String,Int> = mapOf("Ana" to 46, "Luis" to 27, "João" to 15)
val ageOrNull: Int? = personAges["Luis"] // Supostamente, personAges["Luis"] pode ser null
val age = checkNotNull(ageOrNull) { "Age of Luis is missing" }   // Parâmetro do tipo Int?, retorno do tipo Int.
println(age::class.simpleName)
```
