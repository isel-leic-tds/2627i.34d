# Valores e Variáveis

* Kotlin é **fortemente tipificada**.
* Suporta **inferência de tipos** (o tipo é deduzido a partir da atribuição inicial).

```kotlin
var days = 30           // Number of days in a month (type Int inferred)
print("November has ")  // --> November has  (cursor stay at end of line)
println(days)           // --> 30            (cursor moves to the next line)
days = 31               // Change the value of variable
print("December has ")  // --> December has  (cursor stay at end of line)
println(days)           // --> 31  
```

* O tipo inferido é estático (não muda após a atribuição).
* Atribuir um tipo incompatível gera erro de compilação:

```kotlin
var days = 30
print("November has ")
println(days)
days = 31.5               // Erro!
print("December has ")
println(days)
```

# Valores Imutáveis *vs.* Variáveis

* **`var`**: Declara variáveis mutáveis.
* **`val`**: Declara valores imutáveis (apenas leitura após atribuição).

```kotlin
val days = 30
print("November has ")
println(days)
```

* Reatribuir um valor `val` resulta em erro de compilação:

```kotlin
val days = 30
print("November has ")
println(days)
days = 31               // Erro!
print("December has ")
println(days)
```

# Tipos Básicos: Inteiros

* Suportam representações com e sem sinal em tamanhos de 8, 16, 32 e 64 bits.

| Tipo | Dimensão | Mínimo | Máximo |
| ----- | :---: | :---: | :---: |
| **Byte** | 1 byte \= 8 bits | \-128 | 127 |
| **Short** | 2 bytes \= 16 bits | \-32768 | 32767 |
| **Int** | 4 bytes \= 32 bits | \-2147483648 | 2147483647 |
| **Long** | 8 bytes \= 64 bits | $-2^{63}$ | $2^{63}-1 $|
| **UByte** | 1 byte \= 8 bits | 0 | 255 |
| **UShort** | 2 bytes \= 16 bits | 0 | 65535 |
| **UInt** | 4 bytes \= 32 bits | 0 | 4294967295 |
| **ULong** | 8 bytes \= 64 bits | 0 | $2^{64}-1$ |

* Por omissão, números inteiros são inferidos como `Int`.

```kotlin
val numero = 3
print(numero::class.simpleName)   // Imprime o nome do tipo de 'numero'
```

# Tipos Básicos: Long

* Se o valor exceder a capacidade de um `Int`, o tipo inferido passa a ser `Long`.

```kotlin
val numero = 30000000000
print(numero::class.simpleName)
```

* O compilador nunca infere automaticamente tipos menores que `Int` (`Short` ou `Byte`).
* Use o sufixo `L` para forçar o tipo `Long` num literal.

```kotlin
val n1 = 3
val n2 = 3L
println(n1::class.simpleName)
println(n2::class.simpleName)
```

# Tipos Básicos: Short e Byte

* Para `Short` e `Byte`, é necessário especificar o tipo explicitamente:

```kotlin
val n1 = 3
val n2: Short = 3
val n3: Byte = 3
println(n1::class.simpleName)
println(n2::class.simpleName)
println(n3::class.simpleName)
```

# Tipos Básicos: Inteiros Sem Sinal

* Para tipos sem sinal, usam-se os sufixos `U` (`UInt`) e `LU` (`ULong`):

```kotlin
val n1 = 3
val n2 = 3U
val n3 = 3UL
println(n1::class.simpleName)
println(n2::class.simpleName)
println(n3::class.simpleName)
```

* Para `UShort` e `UByte`, declara-se o tipo explicitamente com um literal sem sinal (`U`):

```kotlin
val n1: UShort = 3U
val n2: UByte = 3U
println(n1::class.simpleName)
println(n2::class.simpleName)
```

# Tipos Básicos: Conversões entre Inteiros

* Não há conversão implícita entre tipos inteiros distintos:

```kotlin
val n1 = 3
var n2 = 0L
println(n1::class.simpleName)
println(n2::class.simpleName)
n2 = n1   // Erro de compilação!
println(n2)
```

* Conversões devem ser explícitas usando funções de conversão:

```kotlin
val n1 = 3
var n2 = 0L
println(n1::class.simpleName)
println(n2::class.simpleName)
n2 = n1.toLong()   // Sem erro!
println(n2)
```

# Tipos Básicos: Conversões entre Inteiros (II)

* Funções disponíveis: `toByte()`, `toShort()`, `toLong()`, `toUByte()`, `toUShort()`, `toUInt()`, `toULong()`. Can be called directly on literals.

```kotlin
val n1 = 3.toLong()
var n2 = 0.toUShort()
println(n1::class.simpleName)
println(n2::class.simpleName)
```

# Tipos Básicos: Limites e *Overflow*

* Kotlin não verifica *overflow* de **operações** em tempo de compilação ou execução.

```kotlin
val n1 = 2_000_000_000  // Inferido como Int
println(n1::class.simpleName)
println(n1 * 10)   // Overflow: 20*10^9 é muito grande para um Int
```

* O compilador valida apenas os limites de atrabuição de **literais**:

```kotlin
val n1: Int = 20_000_000_000  // Literal é grande demais para um Int. Erro de compilação.
```

# Tipos Básicos: Operações entre Inteiros

* **Operadores básicos**: `+`, `-`, `*`, `/` (divisão inteira), `%` (resto).

```kotlin
println(5 / 2)
println(5 % 2) 
println(5 + 2)
println(5 - 2)
println(5 * 2)
```

# Tipos Básicos: Operações entre Inteiros (II)

* **Regras de tipo de resultado**:
  * Se envolver `Long`, o resultado será `Long`.
  * Nos restantes casos, o resultado é promovido para `Int`.

```kotlin
val n1: Int = 5
val n2: Int = 3
val n3: Short = 3
val n4: Long = 3
val r1 = n1 / n2
val r2 = n1 / n3
val r3 = n1 / n4
println(r1)
println(r1::class.simpleName)
println(r2)
println(r2::class.simpleName)
println(r3)
println(r3::class.simpleName)
```

# Tipos Básicos: Operações entre Inteiros (III)

* Operações entre `Short` ou `Byte` também resultam em `Int`.

```kotlin
val n1: Short = 5
val n2: Short = 3
val r1 = n1 / n2
println(r1)
println(r1::class.simpleName)
```

* Para tipos sem sinal, aplicam-se as mesmas regras mantendo o sinal desativado no tipo resultante:

```kotlin
val n1: UInt = 5U
val n2: UInt = 3U
val r1 = n1 / n2
println(r1)
println(r1::class.simpleName)
```

# Tipos Básicos: Literais em Outras Bases

* **Prefixos de bases numéricas**:
  * `0b`: Binário
  * `0x`: Hexadecimal

```kotlin
val n1 = 0b1010   // 10 em decimal
val n2 = 0x23     // 35 em decimal
println(n1)
println(n2)
```

# Tipos Básicos: Valores Reais

* Dois tipos suportados: `Float` (32 bits) e `Double` (64 bits).

| Tipo | Dimensão | Mantissa | Expoente | Precisão |
| ----- | :---: | :---: | :---: | :---: |
| **Float** | 4 bytes \= 32 bits | 23 bits | 8 bits | 6 a 7 dígitos |
| **Double** | 8 bytes \= 64 bits | 52 bits | 11 bits | 15 a 16 dígitos |

# Tipos Básicos: Literais Reais

* Literais reais usam ponto (`.`) como separador decimal:

```kotlin
val n1 = 5   // Inteiro
val n2 = 5.0 // Double
println(n1::class.simpleName)
println(n2::class.simpleName)
```

* Suportam notação científica (`e` ou `E`):

```kotlin
val n3 = 2.4e6 // = 2,4 x 10^6
val n4 = 2e3 // = 2 x 10^3
println(n3)
println(n4)
```

# Tipos Básicos: Double *vs.* Float

* O tipo por omissão é `Double`. Use o sufixo `F` para definir como `Float`:

```kotlin
val n5 = 5.0F // Float
println(n5::class.simpleName)
```

# Tipos Básicos: Operações sobre Reais

* **Operadores suportados**: `+`, `-`, `*`, `/`.
* **Tipo do Resultado**: Se incluir `Double`, o resultado é `Double`; caso contrário, é `Float`.

```kotlin
val n1 = 3.0 // Double
val n2 = 5.0F // Float
val n3 = 2e3 // Double
val n4 = 3e2F // Float
val r1 = n1 * n3 // Double
val r2 = n1 * n2 // Double
val r3 = n2 * n4 // Float
println(r1::class.simpleName)
println(r2::class.simpleName)
println(r3::class.simpleName)
```

# Tipos Básicos: Conversões de/para Reais

* Uso das funções explícitas `toDouble()` e `toFloat()`:

```kotlin
val n1 = 3.0 // Double
val n2 = n1.toFloat() // Float

println(n1::class.simpleName)
println(n2::class.simpleName)
```

* A função `toInt()` converte para inteiro por **truncagem** (descarta a parte decimal):

```kotlin
val n1 = 3.7
println(n1.toInt())
```

# Tipos Básicos: Texto

#### Char

* Representa um único caractere Unicode (2 bytes em memória).
* Definido entre aspas simples (`'...'`):

```kotlin
val ch = 'A'
```

* Suporta sequências de escape (`\t`, `\n`, etc.) e código Unicode (`\uXXXX`):

```kotlin
val tab = '\t'                 // Tab
val euro = '\u20AC'            // Unicode para o símbolo €
print("|")
print(tab)
print("|")
print(euro)
```

# Tipos Básicos: Mais sobre Char

* Obtenção do código Unicode via propriedade `.code`:

```kotlin
val cA: Int = 'A'.code
println(cA)
```

* Suporta aritmética direta (`+` e `-`) com base no código Unicode:

```kotlin
val d: Char = 'A' + 3             // Char + Int = Char
val y = 'Z' - 1                   // Char - Int = Char
val dif: Int = 'Y' - 'D'          // Char - Char = Int
println(d)
println(y)
println(dif)  
```

# Tipos Básicos: String

* Cadeia de caracteres (representada em UTF-16 na JVM).
* Delimitada por aspas duplas (`"..."`):

```kotlin
val s1 = "TDS"
val s2 = "T\tD\tS\nISEL"
val s3 = "Preço: \u20AC 3,00"

println(s1)
println(s2)
println(s3)
```

# Tipos Básicos: *String Templates*

* Permite embutir variáveis (`$var`) ou expressões (`${expr}`) diretamente no texto:

```kotlin
val daysOfNov = 30
println("November has $daysOfNov days")       // --> November has 30 days
println("December has ${daysOfNov + 1} days") // --> Dezember has 31 days
```

# Tipos Básicos: *Raw Strings*

* Delimitadas por `"""..."""`.
* Permite preservar quebras de linha e formatação sem uso de escapes:

```kotlin
val rawString = """Uma string longa
com uma quebra de linha
    alguns espaços no início da outra linha.


"""
print(rawString)
```

# Tipos Básicos: *Raw Strings* (II)

* Sequências de escape não são interpretadas em raw strings:

```kotlin
val rawString = """Será possível incluir um tab (\t) nesta string? E uma quebra de linha (\n)?"""
print(rawString)
```

# Tipos Básicos: Acesso a Caracteres em Strings

* **Acesso a caracteres**: Indexação via `[index]` (inicia em 0).
* **Tamanho**: Propriedade `.length`.

```kotlin
val s = "Kotlin"
val first = s[0]                        // Primeiro caractere
val last = s[s.length-1]                // Último caractere
println("$s: first=$first, last=$last")
```

# Tipos Básicos: Concatenação de Strings

* **Concatenação**: Através do operador `+` ou via string templates:

```kotlin
val s = "Kotlin"
val final = '!'
val txt1 = s + " é fixe" + final  // Concatenation: "Kotlin é fixe!"
val txt2 = "$s é fixe$final"      // String template: "Kotlin é fixe!"
println(txt1)
println(txt2)
```

# Tipos Básicos: Conversão para Strings

* **Conversão para String**: Função `.toString()` (suporta radiciação/base em tipos inteiros):

```kotlin
val i = 36
val si = i.toString()
val d = 13.71
val sd = d.toString()
println(si)
println(sd)
println(i.toString(2))
println(i.toString(16))
```

# Tipos Básicos: Conversão a partir de Strings

* **Conversão a partir de String**: Funções como `.toInt()`, `.toDouble()`, etc.:

```kotlin
println("011".toInt() + "011".toByte(2))
println("3.4e4".toDouble() + 1)  
```

# Tipos Básicos: Valores Lógicos

* Tipo `Boolean`: assume `true` ou `false`.

```kotlin
val ok: Boolean = true              
val notOk = false
println("ok = $ok, notOk = $notOk")
```

* **Operadores de comparação**: `==`, `!=`, `<`, `>`, `<=`, `>=` (retornam `Boolean`).
* **Operadores lógicos**: `&&` (AND), `||` (OR).

```kotlin
val max = 36
val b1: Boolean = 5 < max 
val res = b1 || 30 >= max
println(res)
```

# Tipos Básicos: Avaliação Parcial *vs.* Total

* **Avaliação Parcial (`&&`, `||`)**: O segundo operando só é avaliado se necessário.
* **Avaliação Total (`and`, `or`)**: Ambos os operandos são sempre avaliados.

```kotlin
val res1 = false && readln().toBoolean() // readln não é chamada  
val res2 = false and readln().toBoolean() // readln é chamada
```

* A função `.toBoolean()` converte uma `String` para `Boolean` (retorna `true` apenas se a string for exatamente `"true"`).
